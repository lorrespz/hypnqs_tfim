This is part of the work arxiv:2506.25600. This is a training notebook example for Lorentz GRU NQS. 

```python
import sys
import os
sys.path.append('../utility_tfim')
from lorentz_tfim2d_1drnn_train_loop import *
import time
```

```python
def set_cpu_deterministic(seed):
    # 1. Python & Numpy
    random.seed(seed)
    np.random.seed(seed)
    os.environ['PYTHONHASHSEED'] = str(seed)
    
    # 2. PyTorch CPU
    torch.manual_seed(seed)
    
    # 3. Force Deterministic Algorithms
    # This prevents non-deterministic CPU operations (like some views/reductions)
    torch.use_deterministic_algorithms(True)
    
    # 4. Limit CPU Threads
    torch.set_num_threads(1)
```

```python
Nx=12
Ny=12
Jz=np.ones((Nx,Ny))
Bx=3.0

units = 70
numsteps = 1001
nsamples = 80
var_tol=65.0

lr1= 1e-2
lr2=5e-3

E_dmrg = -456.90666
```

# seed 111

```python
seed=111
fname = f'2D_TFIM_results_seed={seed}/({Nx},{Ny})/LorentzGRU'
set_cpu_deterministic(seed)
```

## spatial_clamp=4.0

```python
wf= Lorentzwavefunction(systemsize_x=Nx, systemsize_y=Ny, cell_type='LorentzGRU', 
                        units=units, spatial_clamp=4.0, seed=111)

t1 = time.time()
mE, vE = run_2DTFIM(wf, numsteps=numsteps, Nx=Nx, Ny=Ny, Bx=+3, numsamples=nsamples, 
                    lr1=lr1,lr2=lr2, var_tol=var_tol, seed=111, fname=fname)
t2 = time.time()
print(f'Time taken = {np.round((t2-t1)/3600,3)} hrs')
```
