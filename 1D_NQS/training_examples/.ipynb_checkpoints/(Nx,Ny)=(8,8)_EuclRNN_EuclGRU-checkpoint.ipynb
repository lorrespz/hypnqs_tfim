# 2DTFIM - (8,8): 1D Euclidean RNN/GRU NQS

This is part of the work arxiv:2506.25600 ("Two dimensional Hyperbolic RNN Neural Quantum State" by HL Dao).

```python
import time
import random
import os
import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import sys
sys.path.append('../utility_tfim')
from poincare_tfim2d_1drnn_train_loop import *
```

```python
def set_cpu_deterministic(seed):
    # 1. Python & Numpy
    random.seed(seed)
    np.random.seed(seed)
    os.environ['PYTHONHASHSEED'] = str(seed)
    
    # 2. PyTorch CPU
    torch.manual_seed(seed)
    
    # 3. Force Deterministic Algorithms
    # This prevents non-deterministic CPU operations (like some views/reductions)
    torch.use_deterministic_algorithms(True)
    
    # 4. Limit CPU Threads
    torch.set_num_threads(1)
```

```python
Nx=8
Ny=8
num_units = 60
numsteps = 1001
numsamples = 80
var_tol=25.0
lr1=1e-2
lr2=5e-3

E_dmrg =  -202.5077381261612
```

# seed=333

```python
seed = 333
fname = f'2DTFIM_1DNQS_results_seed={seed}/({Nx},{Ny})/Euclidean'
set_cpu_deterministic(seed)
```

## 1D Euclidean RNN

```python
wf = RNNwavefunction(Nx, Ny, 'EuclRNN', num_units, seed=seed)  
t0 = time.time()
RNNEnergy, varRNNEnergy = run_2DTFIM(wf, numsteps = numsteps, Nx = Nx, Ny = Ny,
                                     Bx = 3, numsamples = numsamples, 
                                     lr1 = lr1, var_tol=var_tol, seed = seed, fname = fname)
t1 = time.time()
print(f'Training took {np.round((t1-t0)/3600,3)} hours')
```

## 1D Euclidean GRU

```python
wf2 = RNNwavefunction(Nx, Ny, 'EuclGRU', num_units, seed=seed)  
t0 = time.time()
RNNEnergy, varRNNEnergy = run_2DTFIM(wf2, numsteps = numsteps, Nx = Nx, Ny = Ny,
                                     Bx = 3, numsamples = numsamples, 
                                     lr1 = lr1, var_tol=var_tol, seed = seed, fname = fname)
t1 = time.time()
print(f'Training took {np.round((t1-t0)/3600,3)} hours')
```
