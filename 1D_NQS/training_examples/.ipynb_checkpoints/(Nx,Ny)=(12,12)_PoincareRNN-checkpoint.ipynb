# 2DTFIM - (12,12): 1D PoincareRNN NQS

This is part of the work arxiv:2506.25600 ("Two dimensional Hyperbolic RNN Neural Quantum State" by HL Dao).

```python
import time
import random
import os
import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import sys
sys.path.append('../utility_tfim')
from poincare_tfim2d_1drnn_train_loop import *
```

```python
def set_cpu_deterministic(seed):
    # 1. Python & Numpy
    random.seed(seed)
    np.random.seed(seed)
    os.environ['PYTHONHASHSEED'] = str(seed)
    
    # 2. PyTorch CPU
    torch.manual_seed(seed)
    
    # 3. Force Deterministic Algorithms
    # This prevents non-deterministic CPU operations (like some views/reductions)
    torch.use_deterministic_algorithms(True)
    
    # 4. Limit CPU Threads
    torch.set_num_threads(1)
```

```python
Nx=12
Ny=12
num_units = 70
numsteps = 1001
numsamples = 80
var_tol=65.0
lr1=1e-2
lr2=5e-3
E_dmrg =-457.04155104
```

# Seed = 222

```python
seed = 222
fname = f'2DTFIM_1DNQS_results_seed_{seed}/({Nx},{Ny})/PoincareRNN'
set_cpu_deterministic(seed)
```

## Rmax=0.8

```python
r_max=0.8
wf3= RNNwavefunction_hyp(Nx, Ny, cell_type='HypRNN',r_max=r_max, bias_geom='hyp',
                           hyp_non_lin='id', units=num_units, seed=seed)
t0 = time.time()
RNNEnergy, varRNNEnergy = run_2DTFIM(wf3, numsteps = numsteps, Nx = Nx, Ny = Ny,
                                     Bx = 3, numsamples = numsamples, 
                                     lr1 = lr1, lr2 = lr2, var_tol=var_tol, seed = seed, fname = fname)
t1 = time.time()
print(f'Training took {np.round((t1-t0)/3600,3)} hours')
```
