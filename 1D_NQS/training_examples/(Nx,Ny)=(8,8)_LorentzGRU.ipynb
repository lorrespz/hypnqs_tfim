# 2DTFIM (8,8): 1D LorentzGRU NQS

This is part of the work arxiv:2506.25600 ("Two dimensional Hyperbolic RNN Neural Quantum State" by HL Dao). For seed 111, Lmax=[2.0, 4.0, 6.0] were tested. Lmax=6.0 was chosen (best performance). For seeds=222,333, only Lmax=6.0 was run. 

```python
import sys
import os
sys.path.append('../utility_tfim')
from lorentz_tfim2d_1drnn_train_loop import *
import time
```

```python
def set_cpu_deterministic(seed):
    # 1. Python & Numpy
    random.seed(seed)
    np.random.seed(seed)
    os.environ['PYTHONHASHSEED'] = str(seed)
    
    # 2. PyTorch CPU
    torch.manual_seed(seed)
    
    # 3. Force Deterministic Algorithms
    # This prevents non-deterministic CPU operations (like some views/reductions)
    torch.use_deterministic_algorithms(True)
    
    # 4. Limit CPU Threads
    torch.set_num_threads(1)
```

```python
Nx=8
Ny=8
Jz=np.ones((Nx,Ny))
Bx=3.0

units = 60
numsteps = 1001
nsamples = 80
var_tol=45.0

lr1= 1e-2
lr2=5e-3
```

# seed=333

```python
seed = 333
fname = f'2DTFIM_1DNQS_results_seed={seed}/({Nx},{Ny})/LorentzGRU'
set_cpu_deterministic(seed)
```

## Lmax=6.0

```python
wf= Lorentzwavefunction(systemsize_x=Nx, systemsize_y=Ny, cell_type='LorentzGRU', 
                        units=units, spatial_clamp=6.0, non_lin='id', seed=seed)

t1 = time.time()
mE, vE = run_2DTFIM(wf, numsteps=numsteps, Nx=Nx, Ny=Ny, Bx=+3, numsamples=nsamples, 
                    lr1=lr1,lr2=lr2, var_tol=var_tol, seed=seed, fname=fname)
t2 = time.time()
print(f'Time taken = {np.round((t2-t1)/3600,3)} hrs')
```
