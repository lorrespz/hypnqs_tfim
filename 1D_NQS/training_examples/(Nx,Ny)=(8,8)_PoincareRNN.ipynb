# 2DTFIM - (8,8): 1D PoincareRNN NQS

This is part of the work arxiv:2506.25600 ("Two dimensional Hyperbolic RNN Neural Quantum State" by HL Dao). For seed=111, Rmax=[0.65, 0.8, 1.0] were tested and Rmax=1.0 was chosen for later runs with seeds=222,333.

```python
import time
import random
import os
import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import sys
sys.path.append('../utility_tfim')
from poincare_tfim2d_1drnn_train_loop import *
```

```python
def set_cpu_deterministic(seed=111):
    # 1. Python & Numpy
    random.seed(seed)
    np.random.seed(seed)
    os.environ['PYTHONHASHSEED'] = str(seed)
    
    # 2. PyTorch CPU
    torch.manual_seed(seed)
    
    # 3. Force Deterministic Algorithms
    # This prevents non-deterministic CPU operations (like some views/reductions)
    torch.use_deterministic_algorithms(True)
    
    # 4. Limit CPU Threads
    torch.set_num_threads(1)
```

```python
Nx=8
Ny=8
num_units = 60
numsteps = 1001
numsamples = 80
var_tol=25.0
lr1=1e-2
lr2=5e-3
#E_dmrg = -316.97705
```

# seed 333

```python
seed=333
set_cpu_deterministic(seed)
fname = f'2DTFIM_1DNQS_results_seed={seed}/({Nx},{Ny})/PoincareRNN'
```

## Rmax=1.0

```python
r_max=1.0
wf3= RNNwavefunction_hyp(Nx, Ny, cell_type='HypRNN',r_max=r_max, bias_geom='hyp',
                           hyp_non_lin='id', units=num_units, seed=seed)
t0 = time.time()
RNNEnergy, varRNNEnergy = run_2DTFIM(wf3, numsteps = numsteps, Nx = Nx, Ny = Ny,
                                     Bx = 3, numsamples = numsamples, 
                                     lr1 = lr1, lr2 = lr2, var_tol=var_tol, seed = seed, fname = fname)
t1 = time.time()
print(f'Training took {np.round((t1-t0)/3600,3)} hours')
```
