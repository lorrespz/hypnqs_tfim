```python
import sys
sys.path.append('../utility_tfim')
from xdrnn_tfim2d_train_loop import *
```

# (Nx, Ny)= (8,8)

```python
nx=8
ny=8
units=50
E_88=-202.5077381
fname=f'2DTFIM_2dNQS_res/{(nx,ny)}/Lorentz'
```

## Lmax=2.0, 'id' activation

```python
spatial_clamp = 2.0
wf=LorentzRNNwavefunction(nx, ny, units, 2, spatial_clamp, 'id',  seed = 111)
#convert to torch.float64
wf.model.double()
print(f'{wf.name} with total params = {sum(p.numel() for p in wf.model.parameters())}')

t0 = time.time()
RNNEnergy, varRNNEnergy = run_2DTFIM(wf, numsteps = 551, Nx = nx, Ny = ny,
                                     Bx = 3,  numsamples = 80, 
                                     lr1 = 1e-2, lr2=5e-3, var_tol = 2.0, seed = 111, fname=fname)
t1 = time.time()
print(f'Training took {np.round((t1-t0)/3600,3)} hours')
```

```python
spatial_clamp = 2.0
wf=LorentzRNNwavefunction(nx, ny, units, 2, spatial_clamp, 'elu',  seed = 111)
#convert to torch.float64
wf.model.double()
print(f'{wf.name} with total params = {sum(p.numel() for p in wf.model.parameters())}')

t0 = time.time()
RNNEnergy, varRNNEnergy = run_2DTFIM(wf, numsteps = 551, Nx = nx, Ny = ny,
                                     Bx = 3,  numsamples = 80, 
                                     lr1 = 1e-2, lr2=5e-3, var_tol = 2.0, seed = 111, fname=fname)
t1 = time.time()
print(f'Training took {np.round((t1-t0)/3600,3)} hours')
```
