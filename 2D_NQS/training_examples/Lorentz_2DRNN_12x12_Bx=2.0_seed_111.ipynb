# Bx=2.0

```python
nx=12
ny=12
units=60
B=2.0
E_dmrg=-346.98277872
fname=f'2DTFIM_2dNQS_res/{(nx,ny)}_B={B}/Lorentz'
```

## Lmax=2.0

```python
spatial_clamp = 2.0
wf=LorentzRNNwavefunction(nx, ny, units, 2, spatial_clamp, 'elu',  seed = 111)
#convert to torch.float64
wf.model.double()
print(f'{wf.name} with total params = {sum(p.numel() for p in wf.model.parameters())}')

t0 = time.time()
RNNEnergy, varRNNEnergy = run_2DTFIM(wf, numsteps = 551, Nx = nx, Ny = ny,
                                     Bx = B,  numsamples = 80, 
                                     lr1 = 1e-2, lr2=5e-3, var_tol = 4.0, seed = 111, fname=fname)
t1 = time.time()
print(f'Training took {np.round((t1-t0)/3600,3)} hours')
```
