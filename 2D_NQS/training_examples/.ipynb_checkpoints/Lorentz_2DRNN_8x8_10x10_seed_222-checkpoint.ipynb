```python
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import sys
sys.path.append('../../utility_tfim')
from xdrnn_tfim2d_train_loop import *
```

```python
seed = 222

def set_cpu_deterministic(seed):
    # 1. Python & Numpy
    random.seed(seed)
    np.random.seed(seed)
    os.environ['PYTHONHASHSEED'] = str(seed)
    
    # 2. PyTorch CPU
    torch.manual_seed(seed)
    
    # 3. Force Deterministic Algorithms
    # This prevents non-deterministic CPU operations (like some views/reductions)
    torch.use_deterministic_algorithms(True)
    
    # 4. Limit CPU Threads
    torch.set_num_threads(1)

set_cpu_deterministic(seed)
```

# (Nx, Ny)= (8,8)

```python
nx=8
ny=8
units=50
E_88=-202.5077381
fname=f'2DTFIM_2dNQS_res/{(nx,ny)}/Lorentz'
```

```python
spatial_clamp = 2.0
wf=LorentzRNNwavefunction(nx, ny, units, 2, spatial_clamp, 'elu',  seed = seed)
#convert to torch.float64
wf.model.double()
print(f'{wf.name} with total params = {sum(p.numel() for p in wf.model.parameters())}')

t0 = time.time()
RNNEnergy, varRNNEnergy = run_2DTFIM(wf, numsteps = 551, Nx = nx, Ny = ny,
                                     Bx = 3,  numsamples = 80, 
                                     lr1 = 1e-2, lr2=5e-3, var_tol = 2.0, seed = seed, fname=fname)
t1 = time.time()
print(f'Training took {np.round((t1-t0)/3600,3)} hours')
```
